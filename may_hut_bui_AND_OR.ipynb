```python
import random

class Node:
    def __init__(self, room, x, y, parent, action, step):
        self.room = room
        self.x = x
        self.y = y
        self.parent = parent
        self.action = action
        self.step = step


m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

start_x = random.randint(0, m - 1)
start_y = random.randint(0, n - 1)


def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("X", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()


def get_rules(x, y):
    actions = []

    if x > 0:
        actions.append("UP")

    if x < m - 1:
        actions.append("DOWN")

    if y > 0:
        actions.append("LEFT")

    if y < n - 1:
        actions.append("RIGHT")

    return actions


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x = x
    new_y = y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


def OR_SEARCH(current_node, path):
    if is_goal(current_node.room):
        return current_node

    state = (room_to_tuple(current_node.room), current_node.x, current_node.y)
    if state in path:
        return None

    for action in get_rules(current_node.x, current_node.y):
        new_room, new_x, new_y = move(current_node.room, current_node.x, current_node.y,action)
        child = Node(new_room, new_x, new_y, current_node, action, current_node.step + 1)
        result_states = [child]
        plan = AND_SEARCH(result_states, path + [state])

        if plan is not None:
            return plan

    return None


def AND_SEARCH(states, path):
    for node in states:
        plan = OR_SEARCH(node,path)
        if plan is None:
            return None

        return plan

    return None


def AND_OR_GRAPH_SEARCH(room, start_x, start_y):

    if room[start_x][start_y] == 1:
        room[start_x][start_y] = 0

    start_node = Node(room, start_x, start_y, None, None, 0)
    return OR_SEARCH(start_node, [])


def print_solution(goal_node):
    path = []
    while goal_node is not None:
        path.append(goal_node)
        goal_node = goal_node.parent

    path.reverse()
    for node in path:
        if node.action is None:
            print(f"Bước {node.step}: Vị trí bắt đầu")
        else:
            print(f"Bước {node.step}: {node.action}")

        print_room(node.room, node.x, node.y)

    print("Tổng số bước:", len(path) - 1)


print("Vị trí bắt đầu:", (start_x, start_y))
print()

result = AND_OR_GRAPH_SEARCH(room, start_x, start_y)

if result:
    print_solution(result)
    print("Phòng đã được làm sạch")
else:
    print("Không tìm thấy lời giải")
```
