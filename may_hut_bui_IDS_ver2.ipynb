```python
import random

class Node:
    def __init__(self, room, parent, action, step):
        self.room = room
        self.parent = parent
        self.action = action
        self.step = step


m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

start_x = random.randint(0, m - 1)
start_y = random.randint(0, n - 1)


def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()


def get_rules(x, y):
    moves = []
    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y
    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


def print_solution(node, start_x, start_y):
    path = []
    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:\n")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)


def DLS2(start_room, start_x, start_y, limit):
    frontier = []
    reached = []
    start_node = Node(start_room, None, None, 0)
    frontier.append((start_node, start_x, start_y))

    while frontier:
        current_node, x, y = frontier.pop()
        if is_goal(current_node.room):
            return current_node

        reached.append((room_to_tuple(current_node.room), x, y))

        if current_node.step >= limit:
            continue

        for act in get_rules(x, y):
            new_room, new_x, new_y = move(current_node.room, x, y, act)
            new_state = (room_to_tuple(new_room), new_x, new_y)

            if new_state not in reached:
                in_frontier = False

                for node, fx, fy in frontier:
                    state = (room_to_tuple(node.room), fx, fy)

                    if state == new_state:
                        in_frontier = True
                        break

                if not in_frontier:
                    new_node = Node(new_room,current_node,act,current_node.step + 1)
                    if is_goal(new_node.room):
                        return new_node
                    frontier.append((new_node, new_x, new_y))

    return None


def IDS2(start_room, start_x, start_y):
    start_room = [row[:] for row in start_room]

    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    for depth in range(100):
        result = DLS2(start_room, start_x, start_y, depth)
        if result:
            return result

    return None


print("Vị trí bắt đầu của máy hút bụi:", (start_x, start_y))
print()

result = IDS2(room, start_x, start_y)

if result:
    print_solution(result, start_x, start_y)
else:
    print("Không tìm được lời giải")
```
