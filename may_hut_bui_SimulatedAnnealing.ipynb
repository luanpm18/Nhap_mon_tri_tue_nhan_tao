```python
import random
import math

class Node:
    def __init__(self, room, parent, action, step, cost):
        self.room = room
        self.parent = parent
        self.action = action
        self.step = step
        self.cost = cost


m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

start_x = random.randint(0, m - 1)
start_y = random.randint(0, n - 1)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def heuristic(room):
    cnt = 0
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                cnt += 1
    return cnt


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves

def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y

def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action} | Cost: {node.cost}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)
    print("Tổng cost:", path[-1].cost)

def SimulatedAnnealing(start_room, start_x, start_y):
    T = 100.0
    T_min = 10
    alpha = 0.95

    current_node = Node(start_room, None, None, 0, heuristic(start_room))
    x = start_x
    y = start_y

    if current_node.room[x][y] == 1:
        current_node.room[x][y] = 0
        current_node.cost = heuristic(current_node.room)

    while T > T_min:
        if current_node.action:
            print(f"Bước {current_node.step}: {current_node.action} | Cost: {current_node.cost}| T: {T:.2f}")
        else:
            print(f"Bước {current_node.step}: Vị trí bắt đầu")
        
        print_room(current_node.room, x, y)

        if is_goal(current_node.room):
            return current_node

        actions = get_rules(x, y)
        if not actions:
            return current_node

        action = random.choice(actions)

        new_room, new_x, new_y = move(current_node.room, x, y, action)
        next_node = Node(new_room, current_node, action, current_node.step + 1, heuristic(new_room))
        delta = next_node.cost - current_node.cost

        if delta < 0:
            current_node = next_node
            x, y = new_x, new_y
        else:
            p = math.exp(-delta / T)
            if random.random() < p:
                current_node = next_node
                x, y = new_x, new_y

        T *= alpha
    
    return current_node

print("Vị trí bắt đầu của máy hút bụi:", (start_x, start_y))
print()

result = SimulatedAnnealing(room, start_x, start_y)

if result and is_goal(result.room):
    print("Đã tìm được lời giải!")
else:
    print("Không tìm được lời giải")
```
