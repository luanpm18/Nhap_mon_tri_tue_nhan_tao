```python
import random

class Node:
    def __init__(self, room, parent, action, step, cost, x, y):
        self.room = room
        self.parent = parent
        self.action = action
        self.step = step
        self.cost = cost
        self.x = x
        self.y = y


m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

start_x = random.randint(0, m - 1)
start_y = random.randint(0, n - 1)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def heuristic(room):
    cnt = 0
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                cnt += 1
    return cnt


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves

def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y

def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action} | Cost: {node.cost}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)
    print("Tổng cost:", path[-1].cost)

def BeamSearch(start_room, start_x, start_y, k=2):
    current_node = Node(start_room, None, None, 0, heuristic(start_room), start_x, start_y)
    x = start_x
    y = start_y

    if current_node.room[x][y] == 1:
        current_node.room[x][y] = 0
        current_node.cost = heuristic(current_node.room)

    beam = [current_node]
    while beam:
        for node in beam:
            if is_goal(node.room):
                return node
        
        candidates = []
        
        for node in beam:
             for act in get_rules(node.x, node.y):
                new_room, new_x, new_y = move(node.room, node.x, node.y, act)
                candidates.append(Node(new_room, node, act, node.step + 1, heuristic(new_room), new_x, new_y))

        if not candidates:
            return None
         
        candidates.sort(key=lambda n: n.cost)
        
        beam = candidates[:k]


print("Vị trí bắt đầu của máy hút bụi:", (start_x, start_y))
print()

result = BeamSearch(room, start_x, start_y, k=2)

if result:
    print_solution(result, start_x, start_y)
    print("Đã tìm được lời giải!")
else:
    print("Không tìm được lời giải")
```
