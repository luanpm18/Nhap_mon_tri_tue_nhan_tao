```python
import random

class Node:
    def __init__(self, room, parent, action, step):
        self.room = room
        self.parent = parent
        self.action = action
        self.step = step


m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

start_x = random.randint(0, m - 1)
start_y = random.randint(0, n - 1)


def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()


def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()

    x, y = start_x, start_y

    print("Các bước làm sạch phòng:")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)


def BFS(start_room, start_x, start_y, add_log=None, update_matrix=None):
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    frontier = []
    reached = set()

    start_node = Node(start_room, None, None, 0)
    frontier.append((start_node, start_x, start_y))

    while frontier:
        current_node, x, y = frontier.pop(0)

        if add_log:
            add_log(f"Lấy ({x},{y}) khỏi frontier")

        if update_matrix:
            update_matrix(current_node.room, x, y)

        if is_goal(current_node.room):
            return current_node

        reached.add((room_to_tuple(current_node.room), x, y))
        if add_log:
            add_log(f"Thêm ({x},{y}) vào reached")

        for act in get_rules(x, y):
            new_room, new_x, new_y = move(current_node.room, x, y, act)
            new_state = (room_to_tuple(new_room), new_x, new_y)

            if new_state not in reached:
                in_frontier = False

                for node, fx, fy in frontier:
                    state = (room_to_tuple(node.room), fx, fy)

                    if state == new_state:
                        in_frontier = True
                        break

                if not in_frontier:
                    new_node = Node(new_room, current_node, act, current_node.step + 1)
                    frontier.append((new_node, new_x, new_y))
                    if add_log:
                        add_log(f"Thêm ({new_x},{new_y}) vào frontier")

    return None


print("Vị trí bắt đầu của máy hút bụi:", (start_x, start_y))
print()

result = BFS(room, start_x, start_y)

if result:
    print_solution(result, start_x, start_y)
else:
    print("Không tìm được lời giải")
```

```python
from collections import deque
import random

m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

def room_to_tuple(room):
    return tuple(tuple(row) for row in room)

def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True

def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def get_rules(x, y):
    moves = []
    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves

def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y
    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1
    
    if room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0
    return new_room, new_x, new_y

def print_solution(node):
    path = []
    while node is not None:
        path.append(node)
        node = node[1]

    path.reverse()
    print("Các bước làm sạch phòng:")
    for room, parent, action, step, x, y in path:
        if action:
            print(f"Bước {step}: {action}")
        else:
            print(f"Bước {step}: Vị trí bắt đầu")

        print_room(room, x, y)

    print("Tổng số bước:", len(path) - 1)

def BFS(start_room, start_x, start_y):
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    frontier = deque()
    reached = []
    start_node = (start_room, None, None, 0, start_x, start_y)
    frontier.append(start_node)
    while frontier:
        current_node = frontier.popleft()
        current_room, parent, action, step, x, y = current_node

        if is_goal(current_room):
            return current_node

        reached.append((room_to_tuple(current_room), x, y))
        rules = get_rules(x, y)

        for act in rules:
            new_room, new_x, new_y = move(current_room, x, y, act)
            new_state = (room_to_tuple(new_room), new_x, new_y)
            if new_state not in reached:
                in_frontier = False
                for node in frontier:
                    state = (room_to_tuple(node[0]), node[4], node[5]) #lấy sơ đồ của sàn nhà và vị trí của máy hút bụi
                    if state == new_state:
                        in_frontier = True
                        break

                if not in_frontier:
                    new_node = (new_room, current_node, act, step + 1, new_x, new_y)
                    frontier.append(new_node)

    return None

print("Vị trí bắt đầu của máy hút bụi:", (x, y))
print()

result = BFS(room, x, y)

if result:
    print_solution(result)
else:
    print("Không tìm được lời giải")
```
