## RULES
### Xét hướng có thể di chuyển
if x > 0 -> UP

if x < m - 1 -> DOWN

if y > 0 -> LEFT

if y < n - 1 -> RIGHT

### Xét đến trạng thái tương lai để lựa chọn đường đi
if action("UP") in visited -> không chọn đường đi up

if action("DOWN") in visited -> không chọn đường đi down

if action("LEFT") in visited -> không chọn đường đi left

if action("RIGHT") in visited -> không chọn đường đi right

### Trạng thái tại một thời điểm của máy hút bụi:
+ vị trí của máy hút bụi

+ môi trường xung quanh máy hút bụi(ô sạch và dơ)

```python
import random

m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

visited = set()

def room_to_tuple(room):
    return tuple(tuple(row) for row in room)

def print_room(room, x, y):
    for i in range(m):
        for j in range(n):

            if i == x and j == y:
                print("R", end=" ")
            else:
                print(room[i][j], end=" ")
        print()

def is_clean(room):
    for row in room:
        if 1 in row:
            return False
    return True

def get_moves(x, y):
    moves = []
    if x > 0:
        moves.append(("UP", x - 1, y))

    if x < m - 1:
        moves.append(("DOWN", x + 1, y))

    if y > 0:
        moves.append(("LEFT", x, y - 1))

    if y < n - 1:
        moves.append(("RIGHT", x, y + 1))

    return moves

def Rules(room, x, y):
    if room[x][y] == 1:
        room[x][y] = 0
        print("Lau chỗ bẩn tại (",x,",",y,")")

    moves = get_moves(x, y)
    safe_moves = []

    for move in moves:
        direction, new_x, new_y = move
        future_state = (new_x, new_y, room_to_tuple(room))

        if future_state not in visited:
            safe_moves.append(move)
        else:
            print("Đường đi", direction, "dẫn đến trạng thái cũ nên không chọn")

    if not safe_moves:
        return None
    choose_move = random.choice(safe_moves)

    return choose_move

step = 0
max_step = 100

while(is_clean(room) == False and step < max_step):
    print("Bước:", step)
    print_room(room, x, y)

    visited.add((x, y, room_to_tuple(room)))
    action = Rules(room, x, y)

    if action == None:
        print("Không còn trạng thái mới để đi")
        break

    direction, x, y = action
    print("Di chuyển:", direction)
    print()

    step += 1

print("Final room:")
print_room(room, x, y)

if(is_clean(room)):
    print("Phòng đã được làm sạch trong", step, "bước.")
elif(step >= max_step):
    print("Không thể làm sạch phòng trong", max_step, "bước.")
else:
    print("Chương trình dừng")
```
