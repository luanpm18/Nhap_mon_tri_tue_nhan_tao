```python
import random

m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))
room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

while room[x][y] == 2:
    x = random.randint(0, m - 1)
    y = random.randint(0, n - 1)

def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("R", end=" ")
            else:
                print(room[i][j], end= " ")
        print()

def is_clean(room):
    for row in room:
        if 1 in row:
            return False
    return True

def get_moves(x, y):
    moves = []
    if x > 0 and room[x - 1][y] != 2:
        moves.append(("UP", x - 1, y))
    if x < m - 1 and room[x + 1][y] != 2:
        moves.append(("DOWN", x + 1, y))
    if y > 0 and room[x][y - 1] != 2:
        moves.append(("LEFT", x, y - 1))
    if y < n - 1 and room[x][y + 1] != 2:
        moves.append(("RIGHT", x, y + 1))

    return moves

def Rules(room, x, y):

    state_val = room[x][y]
    moves = get_moves(x, y)

    if state_val == 1:
        room[x][y] = 0

    choose_move = random.choice(moves)

    return choose_move

step = 0
max_step = 100
while(is_clean(room) == False and step < max_step):
    
    print("Bước:", step)
    print_room(room, x, y)

    if room[x][y] == 1:
        room[x][y] = 0

    action = Rules(room, x, y)

    direction, x, y = action

    print("Di chuyển:", direction)
    print()
    
    step += 1

print("Final room:")
print_room(room, x, y)

if(is_clean(room)):
    print("Phòng đã được làm sạch trong", step, "bước.")
else:
    print("Không thể làm sạch phòng trong", max_step, "bước.")

```
