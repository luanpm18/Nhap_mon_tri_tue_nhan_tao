```python
import random

m = int(input("Nhập số hàng: "))
n = int(input("Nhập số cột: "))

room = []

print("Nhập ma trận:")

for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

x = random.randint(0, m - 1)
y = random.randint(0, n - 1)

def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("R", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def is_clean(room):
    for row in room:
        if 1 in row:
            return False
    return True

def get_moves(x, y):
    moves = []

    if x > 0:
        moves.append(("UP", x - 1, y))

    if x < m - 1:
        moves.append(("DOWN", x + 1, y))

    if y > 0:
        moves.append(("LEFT", x, y - 1))

    if y < n - 1:
        moves.append(("RIGHT", x, y + 1))

    return moves

step = 0
max_steps = 100

while not is_clean(room) and step < max_steps:

    print("Step:", step)
    print_room(room, x, y)

    room[x][y] = 0

    moves = get_moves(x, y)

    chosen_move = random.choice(moves)

    direction, x, y = chosen_move

    print("Action:", direction)
    print()

    step += 1

print("Final Room:")
print_room(room, x, y)

if is_clean(room):
    print("Đã lau sạch")
else:
    print("Không thể lau sạch")
```
