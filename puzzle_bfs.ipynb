### Trạng thái của bảng bao gồm:
+ sơ đồ hiện tại của bảng(status)

+ cha của bảng

+ bước di chuyển của cha để dẫn đến trạng thái hiện tại

+ số bước đi để đtaj được trạng thái hiện tại(lever)

### Cách thức hoạt động:

+ hoạt động dựa trên thuật toán BFS gồm 1 queue frontier để lưu các trạng thái(node)

+ một list reacher để lưu lại các sơ đồ đã xuất hiện để tránh lặp lại

+ ở mỗi bước xét xem node hiện tại có status là goal hay chưa nếu chưa thì đưa tất cả các con của node hiện tại vào queue và tiếp tục duyệt node tiếp theo

```python
from collections import deque

board = []

print("Nhập bảng 3x3:")

for i in range(3):
    row = list(map(int, input().split()))
    board.append(row)

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

def board_to_tuple(board):
    return tuple(tuple(row) for row in board)

def find_zero(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return (i, j)

def is_goal(board):
    return board == goal

def get_rules(state):
    x, y = state
    moves = []
    if x > 0:
        moves.append("UP")

    if x < 2:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < 2:
        moves.append("RIGHT")

    return moves

def move(board, action):
    x, y = find_zero(board)
    new_board = [row[:] for row in board]

    if action == "UP":
        new_board[x][y], new_board[x - 1][y] = new_board[x - 1][y], new_board[x][y]

    elif action == "DOWN":
        new_board[x][y], new_board[x + 1][y] = new_board[x + 1][y], new_board[x][y]

    elif action == "LEFT":
        new_board[x][y], new_board[x][y - 1] = new_board[x][y - 1], new_board[x][y]

    elif action == "RIGHT":
        new_board[x][y], new_board[x][y + 1] = new_board[x][y + 1], new_board[x][y]

    return new_board

def print_board(board):
    for row in board:
        print(row)
    print()

def print_solution(node):
    path = []

    while node is not None:
        path.append(node)
        node = node[1]  # node[1] là cha của node hiện tại

    path.reverse()
    print("Các bước đi để đạt được kết quả: ")
    for board, parent, action, step in path:
        if action:
            print(f"Bước {step}: {action}")
        else:
            print(f"Bước {step}:")
        print_board(board)

    print("Tổng số bước:", len(path) - 1)


def BFS(start_board):
    frontier = deque()
    reached = []

    start_node = (start_board, None, None, 0)
    frontier.append(start_node)

    while frontier:
        current_node = frontier.popleft()
        current_board, parent, action, step = current_node

        if is_goal(current_board):
            return current_node

        reached.append(board_to_tuple(current_board))
        state = find_zero(current_board)
        rules = get_rules(state)

        for act in rules:
            new_board = move(current_board, act)

            if board_to_tuple(new_board) not in reached:
                in_frontier = False
                for node in frontier:
                    if board_to_tuple(node[0]) == board_to_tuple(new_board):
                        in_frontier = True
                        break

                if not in_frontier:
                    new_node = (new_board, current_node, act, step + 1) #curr_node là kiểu như lưu địa chỉ chủa node cha
                    frontier.append(new_node)

    return None


result = BFS(board)

if result:
    print_solution(result)
else:
    print("Không tìm được lời giải")
```
