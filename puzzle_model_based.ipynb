## RULES:
### Xét hướng có thể di chuyển
if x > 0 -> UP

if x < m - 1 -> DOWN

if y > 0 -> LEFT

if y < n - 1 -> RIGHT

### Xét đến trạng thái tương lai để lựa chọn đường đi
if action("UP") in visited -> không chọn đường đi up

if action("DOWN") in visited -> không chọn đường đi down

if action("LEFT") in visited -> không chọn đường đi left

if action("RIGHT") in visited -> không chọn đường đi right

```python
import random

board = []

print("Nhập bảng 3x3:")

for i in range(3):
    row = list(map(int, input().split()))
    board.append(row)

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

visited = set()

def board_to_tuple(board):
    return tuple(tuple(row) for row in board)

def find_zero(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return (i, j)

def is_goal(board):
    return board == goal

def get_rules(state):
    x, y = state
    future_board = [row[:] for row in board]

    moves = []

    if x > 0:
        moves.append("UP")

    if x < 2:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < 2:
        moves.append("RIGHT")

    if "UP" in moves:
        future_board[x][y], future_board[x - 1][y] = future_board[x - 1][y], future_board[x][y]
        if board_to_tuple(future_board) in visited:
            moves.remove("UP")
            print("Đường đi UP dẫn đến trạng thái cũ nên không chọn")
    future_board = [row[:] for row in board]

    if "DOWN" in moves:
        future_board[x][y], future_board[x + 1][y] = future_board[x + 1][y], future_board[x][y]
        if board_to_tuple(future_board) in visited:
            moves.remove("DOWN")
            print("Đường đi DOWN dẫn đến trạng thái cũ nên không chọn")
    future_board = [row[:] for row in board]

    if "LEFT" in moves:
        future_board[x][y], future_board[x][y - 1] = future_board[x][y - 1], future_board[x][y]
        if board_to_tuple(future_board) in visited:
            moves.remove("LEFT")
            print("Đường đi LEFT dẫn đến trạng thái cũ nên không chọn")
    future_board = [row[:] for row in board]

    if "RIGHT" in moves:
        future_board[x][y], future_board[x][y + 1] = future_board[x][y + 1], future_board[x][y]
        if board_to_tuple(future_board) in visited:
            moves.remove("RIGHT")
            print("Đường đi RIGHT dẫn đến trạng thái cũ nên không chọn")

    return moves

def move(board, action):
    x, y = find_zero(board)

    new_board = [row[:] for row in board]

    if action == "UP":
        new_board[x][y], new_board[x - 1][y] = new_board[x - 1][y], new_board[x][y]

    elif action == "DOWN":
        new_board[x][y], new_board[x + 1][y] = new_board[x + 1][y], new_board[x][y]

    elif action == "LEFT":
        new_board[x][y], new_board[x][y - 1] = new_board[x][y - 1], new_board[x][y]

    elif action == "RIGHT":
        new_board[x][y], new_board[x][y + 1] = new_board[x][y + 1], new_board[x][y]

    return new_board


def print_board(board):
    for row in board:
        print(row)
    print()

step = 0
max_steps = 300

while not is_goal(board) and step < max_steps:

    print("Step:", step)
    print_board(board)

    visited.add(board_to_tuple(board))
    all_action = get_rules(find_zero(board))
    action = None

    if not all_action:
        print("Tất cả đường đi đều dẫn đến trạng thái cũ nên dừng chương trình")
        break
    else:
        action = random.choice(all_action)

    print("Action:", action)
    print()

    new_board = move(board, action)

    board = new_board

    step += 1
    print("------------------------------")

print("Final Board:")
print_board(board)

if is_goal(board):
    print("Tìm được lời giải trong", step, "bước")
elif step >= max_steps:
    print("Không tìm được lời giải trong", max_steps, "bước")
else:
    print("Dừng chương trình")
```
