```python
import random

board = []

print("Nhập bảng 3x3:")

for i in range(3):
    row = list(map(int, input().split()))
    board.append(row)

goal = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

visited = set()

def board_to_tuple(board):
    return tuple(tuple(row) for row in board)

def find_zero(board):
    for i in range(3):
        for j in range(3):
            if board[i][j] == 0:
                return (i, j)

def is_goal(board):
    return board == goal

def get_rules(state):
    x, y = state

    moves = []

    if x > 0:
        moves.append("UP")

    if x < 2:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < 2:
        moves.append("RIGHT")

    return moves

def move(board, action):
    x, y = find_zero(board)

    new_board = [row[:] for row in board]

    if action == "UP":
        new_board[x][y], new_board[x - 1][y] = new_board[x - 1][y], new_board[x][y]

    elif action == "DOWN":
        new_board[x][y], new_board[x + 1][y] = new_board[x + 1][y], new_board[x][y]

    elif action == "LEFT":
        new_board[x][y], new_board[x][y - 1] = new_board[x][y - 1], new_board[x][y]

    elif action == "RIGHT":
        new_board[x][y], new_board[x][y + 1] = new_board[x][y + 1], new_board[x][y]

    return new_board

def choose_action(board):
    state = find_zero(board)

    rules = get_rules(state)

    return random.choice(rules)

def print_board(board):
    for row in board:
        print(row)
    print()

step = 0

while not is_goal(board):

    print("Step:", step)
    print_board(board)

    visited.add(board_to_tuple(board))

    action = choose_action(board)

    print("Action:", action)
    print()

    new_board = move(board, action)

    if board_to_tuple(new_board) in visited:
        print("Trạng thái đã xuất hiện trước đó")
        break

    board = new_board

    step += 1

print("Final Board:")
print_board(board)

if is_goal(board):
    print("Tìm được lời giải trong", step, "bước")
else:
    print("Dừng chương trình")
```
