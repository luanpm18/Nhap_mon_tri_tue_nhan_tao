```python
class Node:
    def __init__(self, room, parent, action, step, g = 0, h = 0, cost=0, x=None, y=None):
        self.room = room
        self.parent = parent
        self.action = action
        self.step = step
        self.g = g
        self.h = h
        self.cost = g + h

        self.x = x
        self.y = y

m = 4
n = 4

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

def room_to_string(room):
    return "\n".join(" ".join(map(str, row)) for row in room)


start_x = random.randint(0, 3)
start_y = random.randint(0, 3)
```

```python
#BFS2
def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()

    x, y = start_x, start_y

    print("Các bước làm sạch phòng:")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)


def BFS2(start_room, start_x, start_y, add_log=None, update_matrix=None):
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    frontier = []
    reached = set()

    start_node = Node(start_room, None, None, 0)
    frontier.append((start_node, start_x, start_y))

    while frontier:
        current_node, x, y = frontier.pop(0)

        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y})\n")

        if update_matrix:
            update_matrix(current_node.room, x, y)

        if is_goal(current_node.room):
            return current_node

        reached.add((room_to_tuple(current_node.room), x, y))
        if add_log:
            add_log(f"Thêm vào reached:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y})\n")

        for act in get_rules(x, y):
            new_room, new_x, new_y = move(current_node.room, x, y, act)
            new_state = (room_to_tuple(new_room), new_x, new_y)

            if new_state not in reached:
                in_frontier = False

                for node, fx, fy in frontier:
                    state = (room_to_tuple(node.room), fx, fy)

                    if state == new_state:
                        in_frontier = True
                        break

                if not in_frontier:
                    new_node = Node(new_room, current_node, act, current_node.step + 1)
                    if is_goal(new_node.room):
                        if update_matrix:
                            update_matrix(new_node.room, new_x, new_y)
                        return new_node
                    frontier.append((new_node, new_x, new_y))
                    if add_log:
                        add_log(f"Thêm vào frontier:\n{room_to_string(new_room)}\nVị trí: ({new_x},{new_y})\n")

    return None


```

```python
#IDA*

def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False

    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")

        print()

    print()


def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]

    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    # hút bụi nếu ô mới có bụi
    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


# h(n): số ô bẩn còn lại
def heuristic(room):
    cnt = 0
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                cnt += 1

    return cnt


def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:\n")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action} | g(n) = {node.g} | h(n) = {node.h} | f(n) = {node.cost}")

        else:
            print(f"Bước 0: Vị trí bắt đầu | g(n) = {node.g} | h(n) = {node.h} | f(n) = {node.cost}")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)
    print("Cost cuối cùng:", path[-1].cost)


def get_min_threshold(current_node, x, y, threshold):
    min_threshold = float("inf")

    for act in get_rules(x, y):
        new_room, new_x, new_y = move(current_node.room, x, y, act)

        new_g = current_node.g + 1
        new_h = heuristic(new_room)
        new_cost = new_g + new_h

        if new_cost > threshold:
            if new_cost < min_threshold:
                min_threshold = new_cost

    return min_threshold


def search_with_threshold(start_room, start_x, start_y, threshold, add_log=None, update_matrix=None):

    frontier = []
    reached = {}

    start_g = 0
    start_h = heuristic(start_room)

    start_node = Node(start_room, None, None, 0, start_g, start_h)

    frontier.append((start_node, start_x, start_y))

    if add_log:
        add_log(f"Thêm vào frontier:\n{room_to_string(start_room)}\nVị trí: ({start_x},{start_y})\n")

    min_threshold = float("inf")

    while frontier:
        current_node, x, y = frontier.pop()
        current_state = (room_to_tuple(current_node.room), x, y)

        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n"f"{room_to_string(current_node.room)}\nVị trí: ({x},{y})\ng(n) = {current_node.g} | h(n) = {current_node.h} | f(n) = {current_node.cost}\n")

        if update_matrix:
            update_matrix(current_node.room, x, y)

        if is_goal(current_node.room):
            if add_log:
                add_log("Đã tìm thấy trạng thái đích\n")

            return True, current_node

        reached[current_state] = current_node.cost

        if add_log:
            add_log(f"Thêm vào reached:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y})\n")

        # tìm cost để tăng threshold
        min_threshold_node = get_min_threshold(current_node, x, y, threshold)
        min_threshold = min(min_threshold, min_threshold_node)

        for act in reversed(get_rules(x, y)):
            new_room, new_x, new_y = move(current_node.room, x, y, act)

            new_g = current_node.g + 1
            new_h = heuristic(new_room)

            new_node = Node(new_room, current_node, act, current_node.step + 1, new_g, new_h)

            if new_node.cost > threshold:
                if add_log:
                    add_log(f"Loại bỏ vì vượt threshold:\n{room_to_string(new_room)}\nf(n) = {new_node.cost} > {threshold}\nVị trí: ({x},{y})\ng(n) = {current_node.g} | h(n) = {current_node.h} | f(n) = {current_node.cost}\n")

                continue

            new_state = (room_to_tuple(new_room), new_x, new_y)
            skip = False
            if new_state in reached and reached[new_state] <= new_node.cost:
                skip = True

            for node, fx, fy in frontier:
                state = (room_to_tuple(node.room), fx, fy)

                if state == new_state and node.cost <= new_node.cost:
                    skip = True
                    break

            if not skip:
                frontier.append((new_node, new_x, new_y))
                if add_log:
                    add_log(
                        f"Thêm vào frontier:\n{room_to_string(new_room)}\nVị trí: ({new_x},{new_y})\ng(n) = {new_g} | h(n) = {new_h} | f(n) = {new_node.cost}\n")

    return min_threshold, None


def IDAStar(start_room, start_x, start_y, add_log=None, update_matrix=None):
    start_room = [row[:] for row in start_room]

    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    threshold = heuristic(start_room)

    if add_log:
        add_log(f"Threshold khởi tạo: {threshold}\n")

    while True:
        if add_log:
            add_log(f"\nThreshold = {threshold}\n")

        result, res_node = search_with_threshold(start_room, start_x, start_y, threshold, add_log, update_matrix)

        if result is True:
            if add_log:
                add_log("Tìm thấy lời giải\n")

            return res_node

        if result == float("inf"):
            if add_log:
                add_log("Không tìm thấy lời giải\n")

            return None

        if add_log:
            add_log(f"Tăng threshold từ {threshold} -> {result}\n")

        threshold = result
```

```python
#SimulatedAnnealing

def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def heuristic(room):
    cnt = 0
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                cnt += 1
    return cnt


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves

def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y

def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action} | Cost: {node.cost}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)
    print("Tổng cost:", path[-1].cost)

def SimulatedAnnealing(start_room, start_x, start_y, add_log=None, update_matrix=None):
    T = 100.0
    T_min = 10
    alpha = 0.95

    current_node = Node(start_room, None, None, 0, heuristic(start_room))
    x = start_x
    y = start_y

    if current_node.room[x][y] == 1:
        current_node.room[x][y] = 0
        current_node.cost = heuristic(current_node.room)

    while T > T_min:
        if add_log:
            add_log(f"Trạng thái: {room_to_string(current_node.room)}\nVị trí hiện tại: ({x}, {y}), Cost: {current_node.cost}, T: {T:.2f}\n")
        if update_matrix:
            update_matrix(current_node.room, x, y)

        if is_goal(current_node.room):
            return current_node

        actions = get_rules(x, y)
        if not actions:
            return current_node

        action = random.choice(actions)

        new_room, new_x, new_y = move(current_node.room, x, y, action)
        next_node = Node(new_room, current_node, action, current_node.step + 1, heuristic(new_room))
        delta = next_node.cost - current_node.cost

        if delta < 0:
            current_node = next_node
            x, y = new_x, new_y
        else:
            p = math.exp(-delta / T)
            if random.random() < p:
                current_node = next_node
                x, y = new_x, new_y

        T *= alpha
    
    return current_node
```

```python
#Greedy MTPT
room2 = []

for i in range(m):
    row = [random.randint(0, 1) for _ in range(n)]
    room2.append(row)

start_x2 = random.randint(0, m - 1)
start_y2 = random.randint(0, n - 1)

def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True

def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")

        print()

    print()


def get_rules_mtpt(x, y, x2, y2):
    moves = []
    if x > 0 or x2 > 0:
        moves.append("UP")

    if x < m - 1 or x2 < m - 1:
        moves.append("DOWN")

    if y > 0 or y2 > 0:
        moves.append("LEFT")

    if y < n - 1 or y2 < n - 1:
        moves.append("RIGHT")

    return moves


def move_mtpt(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP" and new_x > 0:
        new_x -= 1

    elif action == "DOWN" and new_x < len(room) - 1:
        new_x += 1

    elif action == "LEFT" and new_y > 0:
        new_y -= 1

    elif action == "RIGHT" and new_y < len(room[0]) - 1:
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y

def cnt_cost(room):
    cnt = 0
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                cnt += 1
    return cnt

def print_solution_mtpt(result, start_x, start_x2, start_y, start_y2):
    node, node2 = result
    path = []
    while node is not None and node2 is not None:
        path.append((node, node2))
        node = node.parent
        node2 = node2.parent

    path.reverse()
    x, y = start_x, start_y
    x2, y2 = start_x2, start_y2
    print("Các bước làm sạch phòng:\n")
    for node, node2 in path:
        if node.action == "STAY":
            pass
        elif node.action:
            _, x, y = move_mtpt(node.parent.room, x, y, node.action)

        if node2.action == "STAY":
            pass
        elif node2.action:
            _, x2, y2 = move_mtpt(node2.parent.room, x2, y2, node2.action)
        
        if node.action is None:
            print("Bước 0: Vị trí bắt đầu")
        else:
            print(f"Bước {node.step}: {node.action} | Cost = {node.cost}")

        print_room(node.room, x, y)
        print_room(node2.room, x2, y2)

    print("Tổng số bước:", len(path) - 1)
    print("Tổng cost:", path[-1][0].cost)

def GreedyMTPT(start_room, start_room2, start_x, start_x2, start_y, start_y2, add_log=None, update_matrix=None):

    start_room = [row[:] for row in start_room]
    start_room2 = [row[:] for row in start_room2]

    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    if start_room2[start_x2][start_y2] == 1:
        start_room2[start_x2][start_y2] = 0

    start_node = Node(start_room, None, None, 0, cnt_cost(start_room))
    start_node2 = Node(start_room2, None, None, 0, cnt_cost(start_room2))

    frontier = [(start_node, start_node2, start_x, start_x2, start_y, start_y2)]
    reached = {}

    while frontier:
        frontier.sort(key=lambda x: max(x[0].cost, x[1].cost))

        current_node, current_node2, x, x2, y, y2 = frontier.pop(0)
        if add_log:
            add_log(f"Trạng thái ban đầu:\n{room_to_string(current_node.room)}\n\n{room_to_string(current_node2.room)}\nVị trí: ({x},{y}), ({x2},{y2})\nCost: {current_node.cost}, {current_node2.cost}\n")

        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n{room_to_string(current_node.room)}\n\n{room_to_string(current_node2.room)}\nVị trí: ({x},{y}), ({x2},{y2})\nCost: {current_node.cost}, {current_node2.cost}\n")

        current_state = (room_to_tuple(current_node.room), room_to_tuple(current_node2.room), x, x2, y, y2)

        reached[current_state] = max(current_node.cost, current_node2.cost)

        if is_goal(current_node.room) and is_goal(current_node2.room):
            return current_node, current_node2

        for act in get_rules_mtpt(x, y, x2, y2):
            if is_goal(current_node.room):
                new_room = current_node.room
                new_x = x
                new_y = y
                new_node = Node(new_room, current_node, "STAY", current_node.step + 1, 0)
            else:
                new_room, new_x, new_y = move_mtpt(current_node.room, x, y, act)
                new_node = Node(new_room, current_node, act, current_node.step + 1, cnt_cost(new_room))

            if is_goal(current_node2.room):
                new_room2 = current_node2.room
                new_x2 = x2
                new_y2 = y2
                new_node2 = Node(new_room2, current_node2, "STAY", current_node2.step + 1, 0)

            else:
                new_room2, new_x2, new_y2 = move_mtpt(current_node2.room, x2, y2, act)
                new_node2 = Node(new_room2, current_node2, act, current_node2.step + 1, cnt_cost(new_room2))

            new_cost = max(cnt_cost(new_room), cnt_cost(new_room2))

            new_state = (room_to_tuple(new_room), room_to_tuple(new_room2), new_x, new_x2, new_y, new_y2)

            skip = False
            if new_state in reached and reached[new_state] <= new_cost:
                skip = True

            for node, node2, fx, fx2, fy, fy2 in frontier:
                state = (room_to_tuple(node.room), room_to_tuple(node2.room), fx, fx2, fy, fy2)
                if state == new_state and max(node.cost, node2.cost) <= new_cost:
                    skip = True
                    break

            if not skip:
                frontier.append((new_node, new_node2, new_x, new_x2, new_y, new_y2))
                if add_log:
                    add_log(f"Thêm vào frontier:\n{room_to_string(new_room)}\n{room_to_string(new_room2)}\nVị trí: ({new_x},{new_y}), ({new_x2},{new_y2})\nCost: {new_node.cost}, {new_node2.cost}\n")

    return None
```

```python
#BFS MTPT
def BFSMTPT(start_room, start_room2, start_x, start_x2, start_y, start_y2, add_log=None, update_matrix=None):
    start_room = [row[:] for row in start_room]
    start_room2 = [row[:] for row in start_room2]

    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    if start_room2[start_x2][start_y2] == 1:
        start_room2[start_x2][start_y2] = 0

    start_node = Node(start_room, None, None, 0, cnt_cost(start_room))
    start_node2 = Node(start_room2, None, None, 0, cnt_cost(start_room2))

    frontier = [(start_node, start_node2, start_x, start_x2, start_y, start_y2)]
    reached = set()
    initial_state = (room_to_tuple(start_room), room_to_tuple(start_room2), start_x, start_x2, start_y, start_y2)
    reached.add(initial_state)

    while frontier:
        current_node, current_node2, x, x2, y, y2 = frontier.pop(0)

        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n{room_to_string(current_node.room)}\n\n{room_to_string(current_node2.room)}\nVị trí: ({x},{y}), ({x2},{y2})\n")

        if is_goal(current_node.room) and is_goal(current_node2.room):
            return current_node, current_node2

        for act in get_rules_mtpt(x, y, x2, y2):
            if is_goal(current_node.room):
                new_room = current_node.room
                new_x = x
                new_y = y
                new_node = Node(new_room, current_node, "STAY", current_node.step + 1, 0)
            else:
                new_room, new_x, new_y = move_mtpt(current_node.room, x, y, act)
                new_node = Node(new_room, current_node, act, current_node.step + 1, cnt_cost(new_room))

            if is_goal(current_node2.room):
                new_room2 = current_node2.room
                new_x2 = x2
                new_y2 = y2
                new_node2 = Node(new_room2, current_node2, "STAY", current_node2.step + 1, 0)
            else:
                new_room2, new_x2, new_y2 = move_mtpt(current_node2.room, x2, y2, act)
                new_node2 = Node(new_room2, current_node2, act, current_node2.step + 1, cnt_cost(new_room2))

            new_state = (room_to_tuple(new_room), room_to_tuple(new_room2), new_x, new_x2, new_y, new_y2)

            if new_state not in reached:
                reached.add(new_state)
                frontier.append((new_node, new_node2, new_x, new_x2, new_y, new_y2))
                if add_log:
                    add_log(f"Thêm vào frontier:\n{room_to_string(new_room)}\n{room_to_string(new_room2)}\nVị trí: ({new_x},{new_y}), ({new_x2},{new_y2})\nCost: {new_node.cost}, {new_node2.cost}\n")

    return None
```

```python
#AND-OR Search
def room_to_tuple(room):
    return tuple(tuple(row) for row in room)

def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True

def get_rules(x, y):
    moves = []
    if x > 0:
        moves.append("UP")
    if x < m - 1:
        moves.append("DOWN")
    if y > 0:
        moves.append("LEFT")
    if y < n - 1:
        moves.append("RIGHT")
    return moves

def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y
    if action == "UP":
        new_x -= 1
    elif action == "DOWN":
        new_x += 1
    elif action == "LEFT":
        new_y -= 1
    elif action == "RIGHT":
        new_y += 1
    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0
    return new_room, new_x, new_y

def AND_OR(start_room, start_x, start_y, add_log=None, update_matrix=None):
    start_room = [row[:] for row in start_room]
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    def OR_SEARCH(current_node, path):
        state = (room_to_tuple(current_node.room), current_node.x, current_node.y)
        if state in path:
            if add_log:
                add_log(f"Phát hiện chu trình tại ({current_node.x},{current_node.y}) -> Quay lui\n")
            return None

        if add_log:
            add_log(f"Đang duyệt trạng thái:\n{room_to_string(current_node.room)}\nVị trí: ({current_node.x},{current_node.y})\n")
        
        if update_matrix:
            update_matrix(current_node.room, current_node.x, current_node.y)

        if is_goal(current_node.room):
            if add_log:
                add_log("Tìm thấy trạng thái đích!\n")
            return current_node

        for action in get_rules(current_node.x, current_node.y):
            new_room, new_x, new_y = move(current_node.room, current_node.x, current_node.y, action)
            child = Node(new_room, current_node, action, current_node.step + 1, x=new_x, y=new_y)
            result_states = [child]
            
            if add_log:
                add_log(f"Thử di chuyển theo hướng {action} đến vị trí ({new_x},{new_y})\n")

            plan = AND_SEARCH(result_states, path + [state])
            if plan is not None:
                return plan

            if add_log:
                add_log(f"Hướng {action} thất bại -> Quay lui về ({current_node.x},{current_node.y})\n")
            if update_matrix:
                update_matrix(current_node.room, current_node.x, current_node.y)

        if add_log:
            add_log(f"Tất cả hướng đi từ ({current_node.x},{current_node.y}) đều thất bại -> Quay lui\n")
        return None

    def AND_SEARCH(states, path):
        for node in states:
            plan = OR_SEARCH(node, path)
            if plan is None:
                return None
            return plan
        return None

    start_node = Node(start_room, None, None, 0, x=start_x, y=start_y)
    return OR_SEARCH(start_node, [])
```
