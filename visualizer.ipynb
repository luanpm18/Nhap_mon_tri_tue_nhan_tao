```python
class Node:
    def __init__(self, room, parent, action, step, g = 0, h = 0, cost=0):
        self.room = room
        self.parent = parent
        self.action = action
        self.step = step
        self.g = g
        self.h = h
        self.cost = g + h

m = 4
n = 4

room = []

print("Nhập ma trận:")
for i in range(m):
    row = list(map(int, input().split()))
    room.append(row)

def room_to_string(room):
    return "\n".join(" ".join(map(str, row)) for row in room)


start_x = random.randint(0, m - 1)
start_y = random.randint(0, n - 1)
```

```python
#BFS2
def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()

def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()

    x, y = start_x, start_y

    print("Các bước làm sạch phòng:")

    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action}")
        else:
            print(f"Bước {node.step}: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)


def BFS2(start_room, start_x, start_y, add_log=None, update_matrix=None):
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    frontier = []
    reached = set()

    start_node = Node(start_room, None, None, 0)
    frontier.append((start_node, start_x, start_y))

    while frontier:
        current_node, x, y = frontier.pop(0)

        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y})\n")

        if update_matrix:
            update_matrix(current_node.room, x, y)

        if is_goal(current_node.room):
            return current_node

        reached.add((room_to_tuple(current_node.room), x, y))
        if add_log:
            add_log(f"Thêm vào reached:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y})\n")

        for act in get_rules(x, y):
            new_room, new_x, new_y = move(current_node.room, x, y, act)
            new_state = (room_to_tuple(new_room), new_x, new_y)

            if new_state not in reached:
                in_frontier = False

                for node, fx, fy in frontier:
                    state = (room_to_tuple(node.room), fx, fy)

                    if state == new_state:
                        in_frontier = True
                        break

                if not in_frontier:
                    new_node = Node(new_room, current_node, act, current_node.step + 1)
                    if is_goal(new_node.room):
                        if update_matrix:
                            update_matrix(new_node.room, new_x, new_y)
                        return new_node
                    frontier.append((new_node, new_x, new_y))
                    if add_log:
                        add_log(f"Thêm vào frontier:\n{room_to_string(new_room)}\nVị trí: ({new_x},{new_y})\n")

    return None


```

```python
#UCS


def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False
    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")

        print()

    print()


def get_rules(x, y):
    moves = []
    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


def print_solution(node, start_x, start_y):
    path = []
    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:\n")
    for node in path:
        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: "f"{node.action} | "f"Cost = {node.cost}")
        else:
            print("Bước 0: Vị trí bắt đầu")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)
    print("Tổng cost:", path[-1].cost)


def UCS(start_room, start_x, start_y, add_log=None, update_matrix=None):
    start_room = [row[:] for row in start_room]
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    frontier = []
    reached = {}
    start_node = Node(start_room, None, None, 0, 0)
    frontier.append((start_node, start_x, start_y))

    start_state = (room_to_tuple(start_room), start_x, start_y)
    reached[start_state] = 0

    while frontier:
        frontier.sort(key=lambda x: x[0].cost)

        current_node, x, y = frontier.pop(0)
        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y}), Cost: {current_node.cost}\n")

        if update_matrix:
            update_matrix(current_node.room, x, y)

        current_state = (room_to_tuple(current_node.room), x, y)
        reached[current_state] = current_node.cost
        if add_log:
            add_log(f"Thêm vào reached:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y}), Cost: {current_node.cost}\n")

        if is_goal(current_node.room):
            return current_node

        for act in get_rules(x, y):
            new_room, new_x, new_y = move(current_node.room, x, y, act)

            if current_node.room[new_x][new_y] == 1:
                new_cost = current_node.cost + 1
            else:
                new_cost = current_node.cost + 3

            new_state = (room_to_tuple(new_room),new_x,new_y)

            skip = False
            if (new_state in reached and reached[new_state] <= new_cost):
                skip = True

            for node, fx, fy in frontier:
                state = (room_to_tuple(node.room), fx, fy)
                if (state == new_state and node.cost <= new_cost):
                    skip = True
                    break

            if (not skip):
                new_node = Node(new_room, current_node, act, current_node.step + 1, new_cost)
                frontier.append((new_node, new_x, new_y))
                if add_log:
                    add_log(f"Thêm vào frontier:\n{room_to_string(new_node.room)}\nVị trí: ({new_x},{new_y}), Cost: {new_node.cost}\n")


    return None
```

```python
#A*

def room_to_tuple(room):
    return tuple(tuple(row) for row in room)


def is_goal(room):
    for row in room:
        if 1 in row:
            return False

    return True


def print_room(room, x, y):
    for i in range(m):
        for j in range(n):
            if i == x and j == y:
                print("x", end=" ")
            else:
                print(room[i][j], end=" ")
        print()
    print()


def get_rules(x, y):
    moves = []

    if x > 0:
        moves.append("UP")

    if x < m - 1:
        moves.append("DOWN")

    if y > 0:
        moves.append("LEFT")

    if y < n - 1:
        moves.append("RIGHT")

    return moves


def move(room, x, y, action):
    new_room = [row[:] for row in room]
    new_x, new_y = x, y

    if action == "UP":
        new_x -= 1

    elif action == "DOWN":
        new_x += 1

    elif action == "LEFT":
        new_y -= 1

    elif action == "RIGHT":
        new_y += 1

    if new_room[new_x][new_y] == 1:
        new_room[new_x][new_y] = 0

    return new_room, new_x, new_y


# h(n): số ô bẩn còn lại
def heuristic(room):
    cnt = 0
    for i in range(m):
        for j in range(n):
            if room[i][j] == 1:
                cnt += 1

    return cnt


def print_solution(node, start_x, start_y):
    path = []

    while node is not None:
        path.append(node)
        node = node.parent

    path.reverse()
    x, y = start_x, start_y
    print("Các bước làm sạch phòng:\n")

    for node in path:

        if node.action:
            _, x, y = move(node.parent.room, x, y, node.action)
            print(f"Bước {node.step}: {node.action} | g(n) = {node.g} | h(n) = {node.h} | f(n) = {node.cost}")
        else:
            print(f"Bước 0: Vị trí bắt đầu | g(n) = {node.g} | h(n) = {node.h} | f(n) = {node.cost}")

        print_room(node.room, x, y)

    print("Tổng số bước:", len(path) - 1)
    print("Cost cuối cùng:", path[-1].cost)


def A_star(start_room, start_x, start_y, add_log=None, update_matrix=None):

    start_room = [row[:] for row in start_room]

    # hút bụi ở vị trí bắt đầu
    if start_room[start_x][start_y] == 1:
        start_room[start_x][start_y] = 0

    frontier = []
    reached = {}

    start_g = 0
    start_h = heuristic(start_room)

    start_node = Node(start_room, None, None, 0, start_g, start_h)
    frontier.append((start_node, start_x, start_y))
    start_state = (room_to_tuple(start_room), start_x, start_y)
    reached[start_state] = start_node.cost

    while frontier:
        frontier.sort(key=lambda x: x[0].cost)

        current_node, x, y = frontier.pop(0)
        if add_log:
            add_log(f"Lấy ra khỏi frontier:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y}), Cost: {current_node.cost}\n")

        if update_matrix:
            update_matrix(current_node.room, x, y)
        
        current_state = (room_to_tuple(current_node.room), x, y)
        if is_goal(current_node.room):
            return current_node

        reached[current_state] = current_node.cost
        if add_log:
            add_log(f"Thêm vào reached:\n{room_to_string(current_node.room)}\nVị trí: ({x},{y}), Cost: {current_node.cost}\n")

        for act in get_rules(x, y):
            new_room, new_x, new_y = move(current_node.room, x, y, act)

            new_g = current_node.g + 1
            new_h = heuristic(new_room)
            new_cost = new_g + new_h

            new_state = (room_to_tuple(new_room), new_x, new_y)
            skip = False

            if new_state in reached and reached[new_state] <= new_cost:
                skip = True

            for node, fx, fy in frontier:
                state = (room_to_tuple(node.room), fx, fy)

                if state == new_state and node.cost <= new_cost:
                    skip = True
                    break

            if not skip:
                new_node = Node(new_room, current_node, act, current_node.step + 1, new_g, new_h)
                frontier.append((new_node, new_x, new_y))
                if add_log:
                    add_log(f"Thêm vào frontier:\n{room_to_string(new_node.room)}\nVị trí: ({new_x},{new_y}), Cost: {new_node.cost}\n")

    return None
```

```python
window = tk.Tk()
window.title("Visualizer for BFS, DFS Algorithms")
window.geometry("1300x800")
is_paused = False

# frame
left_frame = tk.Frame(window, width=150, height=700, bg="lightgray")
left_frame.place(x=0, y=0)

center_frame = tk.Frame(window, width=750, height=700, bg="white")
center_frame.place(x=150, y=0)

right_frame = tk.Frame(window, width=300, height=800, bg="lightgray")
right_frame.place(x=900, y=0)

top_center = tk.Frame(center_frame, width=750, height=500)
top_center.place(x=0, y=0)

bottom_center = tk.Frame(center_frame, width=750, height=200)
bottom_center.place(x=0, y=500)

# log
log_text = tk.Text(right_frame)
log_text.place(x=0, y=0, width=280, height=700)

scroll_log = tk.Scrollbar(right_frame, command=log_text.yview)
scroll_log.place(x=280, y=0, width=20, height=700)

log_text.config(yscrollcommand=scroll_log.set)

# kết quả
label_res = tk.Label(bottom_center, text="Kết quả", font=("Arial", 14))
label_res.place(x=10, y=10)

result_text = tk.Text(bottom_center)
result_text.place(x=0, y=50, width=700, height=250)

scroll_result = tk.Scrollbar(bottom_center, command=result_text.yview)
scroll_result.place(x=700, y=50, width=20, height=250)

result_text.config(yscrollcommand=scroll_result.set)

# matrix
matrix_frame = tk.Frame(top_center)
matrix_frame.place(relx=0.5, rely=0.5, anchor="center")

matrix = []

for i in range(4):
    row = []

    for j in range(4):
        color = "brown" if room[i][j] == 1 else "white"

        cell = tk.Frame(matrix_frame,width=90,height=90,bg=color,bd=1,relief="solid")
        cell.grid(row=i, column=j)

        row.append(cell)

    matrix.append(row)

# hàm
def add_log(text):
    log_text.insert(tk.END, text + "\n")
    log_text.see(tk.END)

def pause():
    global is_paused
    is_paused = True

def resume():
    global is_paused
    is_paused = False

def update_matrix(room, x, y):
    for i in range(4):
        for j in range(4):
            color = "brown" if room[i][j] == 1 else "white"
            matrix[i][j].config(bg=color)

            for widget in matrix[i][j].winfo_children():
                widget.destroy() #xóa robot cũ để không bị chồng lên nhau

            if i == x and j == y:
                label = tk.Label(matrix[i][j], text="O", font=("Arial", 18))
                label.place(relx=0.5, rely=0.5, anchor="center")

    window.update()
    while is_paused:
        window.update()

    window.after(300)

def show_result(node):
    path = []
    while node:
        path.append(node)
        node = node.parent

    path.reverse()
    actions = ["START"]
    for p in path:
        if p.action:
            actions.append(p.action)

    actions.append("END")
    result_text.insert(tk.END, " -> ".join(actions))


def replay_solution(node):
    path = []
    while node:
        path.append(node)
        node = node.parent

    path.reverse()
    x = start_x
    y = start_y

    for p in path:
        if p.action == "UP":
            x -= 1
        elif p.action == "DOWN":
            x += 1
        elif p.action == "LEFT":
            y -= 1
        elif p.action == "RIGHT":
            y += 1
        update_matrix(p.room, x, y)


def run_bfs():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = BFS(new_room,start_x,start_y,add_log=add_log,update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_bfs2():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = BFS2(new_room,start_x,start_y,add_log=add_log,update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_dfs():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = DFS(new_room,start_x,start_y,add_log=add_log,update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_dfs2():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = DFS2(new_room,start_x,start_y,add_log=add_log,update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_ids():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = IDS(new_room,start_x,start_y,add_log=add_log, update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_ids2():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = IDS2(new_room,start_x,start_y,add_log=add_log, update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_ucs():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = UCS(new_room,start_x,start_y,add_log=add_log, update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_greedy():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = Greedy(new_room,start_x,start_y,add_log=add_log, update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

def run_a_star():
    log_text.delete("1.0", tk.END)
    result_text.delete("1.0", tk.END)
    new_room = [row[:] for row in room]
    result = A_star(new_room,start_x,start_y,add_log=add_log, update_matrix=update_matrix)
    if result:
        show_result(result)
        add_log("Đã tìm thấy lời giải")
        add_log("Đang hiển thị đường đi")
        replay_solution(result)

# hiển thị trạng thái ban đầu
update_matrix(room, start_x, start_y)
add_log(f"Vị trí bắt đầu: ({start_x},{start_y})")

#các nút
label_algorithms = tk.Label(left_frame, text="Các thuật toán", bg="lightgray", font=("Arial", 14))
label_algorithms.place(x=10, y=20)

btn_bfs = tk.Button(left_frame, width=10, font=("Arial", 12), text="BFS", command=run_bfs)
btn_bfs.place(x=25, y=130)

btn_bfs2 = tk.Button(left_frame, width=10, font=("Arial", 12), text="BFS2", command=run_bfs2)
btn_bfs2.place(x=25, y=165)

btn_dfs = tk.Button(left_frame, width=10, font=("Arial", 12), text="DFS", command=run_dfs)
btn_dfs.place(x=25, y=200)

btn_dfs2 = tk.Button(left_frame, width=10, font=("Arial", 12), text="DFS2", command=run_dfs2)
btn_dfs2.place(x=25, y=235)

btn_ids = tk.Button(left_frame, width=10, font=("Arial", 12), text="IDS", command=run_ids)
btn_ids.place(x=25, y=270)

btn_ids2 = tk.Button(left_frame, width=10, font=("Arial", 12), text="IDS2", command=run_ids2)
btn_ids2.place(x=25, y=305)

btn_ucs = tk.Button(left_frame, width=10, font=("Arial", 12), text="UCS", command=run_ucs)
btn_ucs.place(x=25, y=340)

btn_greedy = tk.Button(left_frame, width=10, font=("Arial", 12), text="Greedy", command=run_greedy)
btn_greedy.place(x=25, y=375)

btn_a_star = tk.Button(left_frame, width=10, font=("Arial", 12), text="A*", command=run_a_star)
btn_a_star.place(x=25, y=410)

btn_pause = tk.Button(left_frame, width=10, font=("Arial", 12), text="Dừng", command=pause)
btn_pause.place(x=25, y=500)

btn_resume = tk.Button(left_frame, width=10, font=("Arial", 12), text="Tiếp tục", command=resume)
btn_resume.place(x=25, y=535)

window.mainloop()
```
